```python
import numpy as np
import pandas as pd
import os
print(os.getcwd())
```

```python
X_train = np.load('/home/jovyan/work/data/processed/X_train.npy')
X_val   = np.load('/home/jovyan/work/data/processed/X_val.npy'  )
X_test  = np.load('/home/jovyan/work/data/processed/X_test.npy' )
y_train = np.load('/home/jovyan/work/data/processed/y_train.npy')
y_val   = np.load('/home/jovyan/work/data/processed/y_val.npy'  )
y_test  = np.load('/home/jovyan/work/data/processed/y_test.npy' )
```

```python
X_train.shape
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(2)
```

```python
X_train = poly.fit_transform(X_train)
```

```python
# Solution:
X_val = poly.transform(X_val)
X_test = poly.transform(X_test)
```

```python
# Solution:
from sklearn.linear_model import LinearRegression 
```

```python
reg = LinearRegression()
```

```python
reg.fit(X_train, y_train)
```

```python
from joblib import dump  

dump(reg,  '/home/jovyan/work/models/linear_poly_2.joblib')
```

```python
# Solution:
y_train_preds = reg.predict(X_train)
y_val_preds = reg.predict(X_val)
# Solution:
from sklearn.metrics import mean_squared_error as mse
from sklearn.metrics import mean_absolute_error as mae
# Solution:
print(mse(y_train, y_train_preds, squared=False))
print(mae(y_train, y_train_preds))
# Solution:
print(mse(y_val, y_val_preds, squared=False))
print(mae(y_val, y_val_preds))
```
